```python
import pyreadr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "vps_data_subset.rds"   # change path if needed
result = pyreadr.read_r(file_path)
df = result[None].copy()

print(df.shape)
print(df.columns)
df.head()
```

```python
# Datetime
df["DateTimePT"] = pd.to_datetime(df["DateTimePT"], errors="coerce")
df["DateTimeUTC"] = pd.to_datetime(df["DateTimeUTC"], errors="coerce")

# Numeric columns
num_cols = ["Latitude", "Longitude", "HPE", "TempData", "DepthData", "AccelData", "TNL_cm"]
for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Clean sex
df["sex"] = df["sex"].astype(str).str.strip().str.upper()
df["sex"] = df["sex"].replace({
    "MALE": "M",
    "FEMALE": "F"
})

# Date-only for residency
df["DateOnly"] = df["DateTimePT"].dt.date

# Remove rows missing core ID fields only
df = df.dropna(subset=["Tag_shortName", "sex", "HPE"])

# Remove impossible coordinates only for spatial use later
sns.set_theme(style="whitegrid")
```

```python
print("Unique sexes:")
print(df["sex"].value_counts(dropna=False))
print()

print("Non-NA HPE:", df["HPE"].notna().sum())
print("Non-NA TempData:", df["TempData"].notna().sum())
print("Non-NA DepthData:", df["DepthData"].notna().sum())
print("Rows with both Temp and Depth:", df[["TempData", "DepthData"]].notna().all(axis=1).sum())
```

```python
temp_df = df.dropna(subset=["TempData"]).copy()

male_temp_df = temp_df[temp_df["sex"] == "M"].copy()
female_temp_df = temp_df[temp_df["sex"] == "F"].copy()

print("Male temp rows:", male_temp_df.shape)
print("Female temp rows:", female_temp_df.shape)
```

```python
depth_df = df.dropna(subset=["DepthData"]).copy()

male_depth_df = depth_df[depth_df["sex"] == "M"].copy()
female_depth_df = depth_df[depth_df["sex"] == "F"].copy()

print("Male depth rows:", male_depth_df.shape)
print("Female depth rows:", female_depth_df.shape)
```

```python
space_df = df.dropna(subset=["Latitude", "Longitude"]).copy()
space_df = space_df[(space_df["Latitude"] != 0) & (space_df["Longitude"] != 0)]

male_space_df = space_df[space_df["sex"] == "M"].copy()
female_space_df = space_df[space_df["sex"] == "F"].copy()

print("Male spatial rows:", male_space_df.shape)
print("Female spatial rows:", female_space_df.shape)
```

```python
residency_df = df.dropna(subset=["DateOnly"]).copy()

male_res_df = residency_df[residency_df["sex"] == "M"].copy()
female_res_df = residency_df[residency_df["sex"] == "F"].copy()

print("Male residency rows:", male_res_df.shape)
print("Female residency rows:", female_res_df.shape)
```

```python
male_temp_order = male_temp_df["Tag_shortName"].value_counts().index.tolist()
male_depth_order = male_depth_df["Tag_shortName"].value_counts().index.tolist()
male_res_order = male_res_df["Tag_shortName"].value_counts().index.tolist()
```

```python
female_temp_order = female_temp_df["Tag_shortName"].value_counts().index.tolist()
female_depth_order = female_depth_df["Tag_shortName"].value_counts().index.tolist()
female_res_order = female_res_df["Tag_shortName"].value_counts().index.tolist()
```

```python
male_daily_counts = (
    male_res_df.groupby(["Tag_shortName", "DateOnly"])
    .size()
    .reset_index(name="detections")
)

male_heat = male_daily_counts.pivot(
    index="Tag_shortName",
    columns="DateOnly",
    values="detections"
).fillna(0)

male_heat = male_heat.loc[male_res_order]

plt.figure(figsize=(18, max(6, len(male_heat) * 0.3)))
sns.heatmap(
    male_heat,
    cmap="viridis",
    cbar_kws={"label": "Daily detections"}
)
plt.xlabel("Date")
plt.ylabel("Tag_shortName")
plt.title("Male: Daily Detection Heatmap")
plt.tight_layout()
plt.show()
```

```python
female_daily_counts = (
    female_res_df.groupby(["Tag_shortName", "DateOnly"])
    .size()
    .reset_index(name="detections")
)

female_heat = female_daily_counts.pivot(
    index="Tag_shortName",
    columns="DateOnly",
    values="detections"
).fillna(0)

female_heat = female_heat.loc[female_res_order]

plt.figure(figsize=(18, max(6, len(female_heat) * 0.3)))
sns.heatmap(
    female_heat,
    cmap="viridis",
    cbar_kws={"label": "Daily detections"}
)
plt.xlabel("Date")
plt.ylabel("Tag_shortName")
plt.title("Female: Daily Detection Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Use only rows that have coordinates + temperature
male_space_temp = male_space_df.dropna(subset=["TempData"]).copy()
female_space_temp = female_space_df.dropna(subset=["TempData"]).copy()

# Pick top 2 tags WITH usable temp data
male_top2_temp = (
    male_space_temp["Tag_shortName"]
    .value_counts()
    .head(2)
    .index
    .tolist()
)

female_top2_temp = (
    female_space_temp["Tag_shortName"]
    .value_counts()
    .head(2)
    .index
    .tolist()
)

print("Male temp tags:", male_top2_temp)
print("Female temp tags:", female_top2_temp)
```

```python
male_temp_subset = male_space_temp[
    male_space_temp["Tag_shortName"].isin(male_top2_temp)
].copy()

female_temp_subset = female_space_temp[
    female_space_temp["Tag_shortName"].isin(female_top2_temp)
].copy()
```

```python
male_temp_subset['Longitude']
```

```python
plt.figure(figsize=(9, 8))

# Define marker styles for each individual
markers = ["D", "^"]   # circle, triangle (add more if needed)

for i, tag in enumerate(male_top2_temp):
    sub = male_temp_subset[male_temp_subset["Tag_shortName"] == tag]
    
    sc = plt.scatter(
        sub["Longitude"],
        sub["Latitude"],
        c=sub["TempData"],
        cmap="plasma",          # keep consistent color scale
        marker=markers[i],      # different shape per individual
        s=30,
        alpha=0.7,
        label=tag
    )

# One colorbar for temperature
plt.colorbar(sc, label="Temperature")

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Male: Horizontal Space Use (2 Individuals, colored by Temperature)")
plt.legend(title="Tag ID")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 8))

markers = ["D", "^"]   # circle, triangle (add more if needed)

for i, tag in enumerate(female_top2_temp):
    sub = female_temp_subset[female_temp_subset["Tag_shortName"] == tag]
    
    plt.scatter(
        sub["Longitude"],
        sub["Latitude"],
        c=sub["TempData"],
        cmap="plasma",          # keep consistent color scale
        marker=markers[i],      # different shape per individual
        s=30,
        alpha=0.7,
        label=tag
    )

plt.colorbar(label="Temperature")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Female: Horizontal Space Use (2 Individuals, colored by Temperature)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 8), sharex=True, sharey=True)

markers = ["D", "^"]

# -----------------------
# MALE PLOT
# -----------------------
for i, tag in enumerate(male_top2_temp):
    sub = male_temp_subset[male_temp_subset["Tag_shortName"] == tag]
    
    sc = axes[0].scatter(
        sub["Longitude"],
        sub["Latitude"],
        c=sub["TempData"],
        cmap="plasma",
        marker=markers[i],
        s=30,
        alpha=0.7,
        label=tag
    )

axes[0].set_title("Male")
axes[0].set_xlabel("Longitude")
axes[0].set_ylabel("Latitude")
axes[0].legend(title="Tag ID")


# -----------------------
# FEMALE PLOT
# -----------------------
for i, tag in enumerate(female_top2_temp):
    sub = female_temp_subset[female_temp_subset["Tag_shortName"] == tag]
    
    axes[1].scatter(
        sub["Longitude"],
        sub["Latitude"],
        c=sub["TempData"],
        cmap="plasma",
        marker=markers[i],
        s=30,
        alpha=0.7,
        label=tag
    )

axes[1].set_title("Female")
axes[1].set_xlabel("Longitude")
axes[1].legend(title="Tag ID")


# -----------------------
# SHARED COLORBAR
# -----------------------
cbar = fig.colorbar(sc, ax=axes, orientation="vertical")
cbar.set_label("Temperature")


# -----------------------
# FINAL TOUCHES
# -----------------------
fig.suptitle("Horizontal Space Use (2 Individuals per Sex, Colored by Temperature)", fontsize=14)

plt.tight_layout()


# -----------------------
# SAVE FIGURE
# -----------------------
plt.savefig("horizontal_space_use_temp_male_female.png", dpi=300, bbox_inches="tight")

plt.show()
```
