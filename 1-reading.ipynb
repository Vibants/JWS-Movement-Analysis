For my VPS analysis of JWS spatial behavior, thermal habitat use, and combining historical VPS data from the southern California region for comparison, there are plenty good practices that I can and will adopt for efficient data management. The most prominent, and at this point standard in environmental science, strategy is processing my raw VPS detection datasets to create tidy ready data frames where each row represents a detection or position estimate and each column represents variables. Those variables will include temperature, depth, Horizontal Position Error (HPE), ID, sex, size, coordinates, etc. (pg 5).
